# Watercycles 
Watercycles are composed of all the daily flows and transformations of water through a certain space and time. Watercycles are a tool to visualise CWatM outputs, illustrating the balance of inflows, outflows, and storage changes as experienced through a specific space and time. 

*This tool is suggested for simulations where (cells x days) < 50 million*

#### CWatM is to be run including the following settings:

* useSmallLakes = False
* **OUT_MAP_TotalEnd** = cellArea
